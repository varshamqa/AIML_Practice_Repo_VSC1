```python
def ercal(W1, W2, b, gate):
  x1=[0,0,1,1]
  x2=[0,1,0,1]
  output=[]
  error=[]
  for i in range(0,4):
    value =(x1[i]*W1)+ (x2[i]*W2) + b
    if value >=0:
      current_output = 1
    else:
      current_output = 0
    output.append(current_output)

    # Refactored to return the signed error (target - output)
    error_val = gate[i] - current_output
    error.append(error_val)

  print('x1=', x1)
  print('x2=', x2)
  print('y=', gate)
  print('output=', output)
  print('error=', error)
  print('W1=', W1 ,'W2=', W2, 'b=', b)
  print("----------------------")
  return error
```

```python
def PLA(gate):
  c=0
  x1=[0,0,1,1]
  x2=[0,1,0,1]
  b=0
  W1 = 0.2
  W2= 0.2
  learning_rate = 0.1

  error = ercal(W1, W2, b, gate)

  # Increased the maximum number of epochs to allow for convergence
  while sum(list(map(abs, error)))!=0 and c<=50:
    for i in range(0,4):
      # If there is an error, update weights based on the signed error
      if error[i] > 0: # Target was 1, output was 0 (error = 1)
        W1 = W1 + (x1[i]*learning_rate)
        W2 = W2 + (x2[i]*learning_rate)
        b = b + learning_rate
        break # Update based on first error, then re-evaluate all errors
      elif error[i] < 0: # Target was 0, output was 1 (error = -1)
          W1 = W1 - (x1[i]*learning_rate) # Subtract x_i * learning_rate
          W2 = W2 - (x2[i]*learning_rate) # Subtract x_i * learning_rate
          b = b - learning_rate # Subtract learning_rate from bias
          break # Update based on first error, then re-evaluate all errors

    error = ercal(W1, W2, b, gate) # Recalculate errors with updated weights
    c+=1
```

```python
PLA([0,0,0,1]) #AND Gate
```

```python
PLA([0,1,1,1]) # OR Gate
```

```python
PLA([0,1,1,0]) # XOR Gate
```

```python
PLA([1,1,1,0]) # NAND Gate
```

```python
PLA([1,0,0,0]) # NOR Gate
```

```python
PLA([1,0,0,1]) # XNOR Gate
```
